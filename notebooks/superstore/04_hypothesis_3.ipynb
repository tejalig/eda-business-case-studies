#### Hypothesis 3: There has been an increase in discount provided across products, contributing to lower Median Order Value and inconsistent revenue growth.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# loading the data
data = pd.read_parquet("../../data/processed/superstore/data.parquet")
orders = pd.read_parquet("../../data/processed/superstore/orders.parquet")
orders_agg = pd.read_parquet("../../data/processed/superstore/orders_agg.parquet")
dates = pd.read_parquet("../../data/processed/superstore/dates.parquet")
product = pd.read_parquet("../../data/processed/superstore/product.parquet")
product_price_disc = pd.read_parquet("../../data/processed/superstore/product_price_disc.parquet")
```

#### Overall discount trend

```python
# determing whether the typical discount offered has changed over time
data[["Order ID","Discount"]].merge(
    orders[["Order ID","Order Date"]]
).merge(
    dates[["date","year"]],
    left_on="Order Date",
    right_on="date"
    ).groupby("year")["Discount"].agg(
        ["mean","median"]
    ).round(2)
```

- The typical discount value increased after 2015. It went from ~0.31 to ~0.31.5
- The discount increased by bit, the revenue growth slowed. Therefore, potentially reducing the revenue generated from each order and contributing to declining MOV

#### Discount by Category

```python
# discount by Category - MEAN
data[["Order ID","Discount"]].merge(
    orders[["Order ID","Order Date","Product ID"]]
).merge(
    dates[["date","year"]],
    left_on="Order Date",
    right_on="date"
    ).merge(
    product[["Product ID", "Category"]],
    left_on="Product ID", right_on="Product ID"
    ).groupby(
        ["year","Category"]
        )["Discount"].agg(
            ['mean','median']
).reset_index().pivot(
    columns="year",
    index="Category",
    values="mean"
).round(2)
```

```python
# discount by Category - MEDIAN
data[["Order ID","Discount"]].merge(
    orders[["Order ID","Order Date","Product ID"]]
).merge(
    dates[["date","year"]],
    left_on="Order Date",
    right_on="date"
    ).merge(
    product[["Product ID", "Category"]],
    left_on="Product ID", right_on="Product ID"
    ).groupby(
        ["year","Category"]
        )["Discount"].agg(
            ['mean','median']
).reset_index().pivot(
    columns="year",
    index="Category",
    values="median"
).round(2)
```

##### Furniture
- Median discount remained cosnsitent as of 205 throught out the analysis period.
- Mean discount fluctuated between 16% and 17%
- Both measures indicate that the discount distribution for furniture remained highly stable over time, with minimal year-to-year variation
- The company's pricing strategy for Furniture remained consistent throughout the four-year period. 
    - Since both the typical discount and the average discount changed very little, it is unlikely that discounting played a significant role in changes to Furniture's revenue performance.

##### Office supplies
- Median discount decreased from 15% to 0% after 2015.
- Mean discount remained close to 15%.
- The decline in the median indicates that the typical Office Supplies transaction became less likely to receive a discount after 2015. 
    - However, the relatively stable mean suggests that higher discounts were still applied to a smaller subset of transactions, keeping the overall average discount nearly unchanged.
- It suggests the company became more selective about discounting.
    - Instead of giving moderate discounts to many customers, it likely offered larger discounts to a smaller group of transactions while selling most orders at full price.

##### Technology
- Median discount remained constant at 20%.
- Mean discount stayed around 14–15%.
- Both statistics show very little variation, indicating a stable discounting pattern throughout the analysis period.
- Technology maintained a very consistent discounting strategy throughout the analysis period.



#### Discount Distribution

```python
# were discounts more frequent or simply became larger
data[["Order ID","Discount"]].merge(
    orders[["Order ID","Order Date"]]
).merge(
    dates[["date","year"]],
    left_on="Order Date",
    right_on="date"
    ).groupby("year")["Discount"].describe().round(2)
```

- Mean: The mean discount remained between 15% and 16% throughout the four-year period.
    - The average level of discounting was highly stable with negligible year-to-year variation.
-SD: The standard deviation remained approximately 0.20 each year.
    - The variability of discounts remained consistent, suggesting that the spread of discount values did not materially change over time.
- Quartiles: 
    - 25th Percentile = 0%:
        - At least 25% of all transactions received no discount.
    - Median = 50%:
        - The typical discount offered changed slightly in 2016 and 2017, indicating that a greater proportion of transactions received little or no discount compared with earlier years.
    - 75th Percentile = 20%:
        - At least 75% of all transactions received discounts of 20% or less.
- Maximum: Maximum discount remained 80% every year.
    - The highest promotional discounts offered by the company did not change throughout the analysis period.


Things to note:
- The distribution is quite stable. Nothing suggests that the company suddenly became more aggresive with discounting

Therefore, the company's discounting strategy remained largely consistent throughout the analysis period. 
Although, the typical discount changed modestly in 2016 & 2017, the overall average discount, discount variability, upper quartiles and maximum discount remained stable. 

These findings suggest that the business did not rely on increasinlgy aggressive discounting to drive sales.

#### Revenue vs Discount

```python
# determining whether higher discounts are associated with lower revenue
rev_vs_discount = data[["Order ID","Discount","Sales"]].groupby("Discount").agg(
    Revenue = ("Sales","sum"),
    Orders = ("Order ID","nunique")
).reset_index().sort_values("Discount", ascending=True)
rev_vs_discount["Revenue_Per_Order"] = (rev_vs_discount["Revenue"] / rev_vs_discount["Orders"]).round(2)
rev_vs_discount
```

- Most customer orders occurred at 0% and 20% discount levels.
    - 0% discount: 2,644 orders
    - 20% discount: 2,407 orders
    - Together, these two discount levels account for the majority of transactions in the dataset.

- The highest total revenue was generated from:
    - 0% discount: approximately $1.09 million
    - 20% discount: approximately $765 thousand
    - Revenue decreases substantially for higher discount levels, although these discount levels also have considerably fewer orders.

- It can also be obsereved that revenue per order increased.
    - Larger discounts may have been applied to higher-priced products or larger customer orders.
    - Without considering product prices, we cannot conclude that larger discounts improve order value.

The results suggest:

- The majority of company revenue was generated from transactions receiving either no discount or a standard 20% discount, reflecting the organization's primary pricing strategy.
- Although some higher-discount transactions generated relatively high revenue per order, these represented a very small proportion of total orders. 
- This suggests that substantial discounts were likely applied selectively to specific products or high-value transactions rather than being a widespread pricing strategy.
